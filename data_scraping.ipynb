```python
from better_bing_image_downloader import downloader

downloader(
    query="kebab shashlik food",
    limit=100,
    output_dir="/content/kebab_images",
    engine="bing",            # Bing works reliably from Google Colab IPs
    force_replace=True,       # Ensures fresh download attempt
    timeout=60
)
```

```python

from google.colab import drive
drive.mount('/content/drive')
```

```python
import os, shutil

source_dir = '/content/kebab_images/kebab shashlik food'
target_dir = '/content/drive/MyDrive/food101_dataset/kebab'

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
    src_file = os.path.join(source_dir, filename)
    dst_file = os.path.join(target_dir, filename)
    shutil.move(src_file, dst_file)
```
